# ML Training Metrics Visualizer

This notebook demonstrates two ways of visualizing machine learning training metrics:
1. **TensorBoard**: Interactive dashboards for live monitoring
2. **Matplotlib**: Static, publication-ready plots

## Prerequisites

For full functionality, install TensorBoard:
```bash
pip install tensorboard
```

To view TensorBoard dashboard:
```bash
tensorboard --logdir=logs
```

## Import Libraries and Setup

````python
import matplotlib.pyplot as plt
import os
import time

# Try to import TensorBoard, but handle the case where it's not installed
try:
    from torch.utils.tensorboard import SummaryWriter
    TENSORBOARD_AVAILABLE = True
    print("✅ TensorBoard is available for logging.")
except ImportError:
    TENSORBOARD_AVAILABLE = False
    print("❌ TensorBoard is not available. Only Matplotlib plotting will be used.")
    print("💡 To enable TensorBoard logging, install it with: pip install tensorboard")
````

## Matplotlib Plotting Function

This function creates static, publication-ready plots using Matplotlib.

````python
def plot_training_results(train_loss, val_loss, title="Training and Validation Loss", save_path=None):
    """
    Plots and optionally saves the training and validation loss curves using Matplotlib.
    
    Args:
        train_loss (list): A list of floating-point numbers representing the training loss per epoch.
        val_loss (list): A list of floating-point numbers representing the validation loss per epoch.
        title (str): The title for the chart.
        save_path (str, optional): The file path to save the plot image (e.g., 'loss_curve.png').
                                   If set to None, the plot will be displayed directly.
    """
    # 1. Determine the number of epochs from the length of the training loss list.
    epochs = range(1, len(train_loss) + 1)
    
    # 2. Create a new plot figure with a specific size for clarity.
    plt.figure(figsize=(10, 6))
    
    # 3. Plot the training loss data with blue line and circle markers.
    plt.plot(epochs, train_loss, 'bo-', label='Training Loss', linewidth=2, markersize=6)
    
    # 4. Plot the validation loss data with red line and circle markers.
    plt.plot(epochs, val_loss, 'ro-', label='Validation Loss', linewidth=2, markersize=6)
    
    # 5. Add essential plot elements.
    plt.title(title, fontsize=16, fontweight='bold')
    plt.xlabel('Epochs', fontsize=12)
    plt.ylabel('Loss', fontsize=12)
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    
    # 6. Handle output - save or display.
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        plt.close()  # Close the figure to free memory
        print(f"📊 Plot saved to: {save_path}")
    else:
        plt.show()
````

## CSV Logging Function (Fallback)

When TensorBoard is not available, we can save metrics to CSV files for later analysis.

````python
def log_to_csv(train_loss, val_loss, filename="training_log.csv"):
    """
    Alternative logging method that saves metrics to a CSV file.
    This can be used when TensorBoard is not available.
    
    Args:
        train_loss (list): Training loss values
        val_loss (list): Validation loss values
        filename (str): CSV filename to save the data
    """
    import csv
    
    with open(filename, 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(['Epoch', 'Training_Loss', 'Validation_Loss'])
        
        for epoch, (t_loss, v_loss) in enumerate(zip(train_loss, val_loss), 1):
            writer.writerow([epoch, t_loss, v_loss])
    
    print(f"📁 Training metrics saved to: {filename}")
````

## Create Mock Training Data

Let's create some sample data that demonstrates overfitting - where training loss continues to decrease but validation loss starts increasing after epoch 6.

````python
# Create mock data representing overfitting scenario
mock_training_loss = [2.5, 1.8, 1.2, 0.9, 0.7, 0.5, 0.4, 0.3, 0.2]
mock_validation_loss = [2.7, 2.1, 1.6, 1.3, 1.1, 1.0, 1.1, 1.2, 1.4]

print("📈 Mock training data created:")
print(f"Training epochs: {len(mock_training_loss)}")
print(f"Training loss range: {min(mock_training_loss):.1f} - {max(mock_training_loss):.1f}")
print(f"Validation loss range: {min(mock_validation_loss):.1f} - {max(mock_validation_loss):.1f}")
print("\n💡 This data shows overfitting after epoch 6 (validation loss increases while training loss decreases)")
````

## Initialize Logging

Set up logging based on TensorBoard availability.

````python
print("=== ML Training Visualizer Demo ===\n")

# Initialize logging based on TensorBoard availability
writer = None
log_dir = None

if TENSORBOARD_AVAILABLE:
    # Create unique log directory with timestamp
    log_dir = os.path.join("logs", "run_" + str(int(time.time())))
    writer = SummaryWriter(log_dir)
    print(f"📁 TensorBoard log directory: {log_dir}")
else:
    print("📝 Using CSV logging as fallback...")
````

## Simulate Training Loop

This simulates a typical machine learning training loop, logging metrics at each epoch.

````python
# Simulate training loop
print("🚀 Simulating training process...\n")

for epoch in range(len(mock_training_loss)):
    train_loss = mock_training_loss[epoch]
    val_loss = mock_validation_loss[epoch]
    
    # Log to TensorBoard if available
    if TENSORBOARD_AVAILABLE and writer:
        writer.add_scalar('Loss/train', train_loss, epoch + 1)
        writer.add_scalar('Loss/validation', val_loss, epoch + 1)
        writer.add_scalars('Loss/comparison', {
            'train': train_loss, 
            'validation': val_loss
        }, epoch + 1)
    
    print(f"Epoch {epoch+1:2d}: Train Loss={train_loss:.2f}, Val Loss={val_loss:.2f}")

print("\n✅ Training simulation complete!")
````

## Finalize Logging

Close TensorBoard writer or save to CSV as appropriate.

````python
# Close TensorBoard writer if it was used
if writer:
    writer.close()
    print(f"📊 TensorBoard logging complete. View with: tensorboard --logdir={log_dir}")
else:
    # Log to CSV as alternative
    log_to_csv(mock_training_loss, mock_validation_loss, "training_metrics.csv")
````

## Create Static Matplotlib Plot

Generate a publication-ready static plot and save it to file.

````python
# Create static matplotlib plot
print("🎨 Generating static Matplotlib plot...")
plot_training_results(
    mock_training_loss, 
    mock_validation_loss, 
    title="Training vs Validation Loss (Overfitting Example)",
    save_path="final_loss_curve.png"
)
````

## Display Interactive Plot

Show the plot directly in the notebook for immediate visualization.

````python
# Also display the plot
print("📈 Displaying interactive plot...")
plot_training_results(
    mock_training_loss, 
    mock_validation_loss, 
    title="Training vs Validation Loss (Overfitting Example)"
)
````

## Summary

Display a summary of all files created during the demonstration.

````python
print("\n=== Demo Complete ===\n")
print("📁 Files created:")
print("   📊 final_loss_curve.png (static plot)")

if not TENSORBOARD_AVAILABLE:
    print("   📝 training_metrics.csv (metrics data)")
    
if log_dir:
    print(f"   📈 {log_dir}/ (TensorBoard logs)")
    print(f"\n💡 To view TensorBoard: tensorboard --logdir={log_dir}")

print("\n🎉 Visualization demo completed successfully!")
````

## Key Takeaways

### TensorBoard vs Matplotlib

| Feature | TensorBoard | Matplotlib |
|---------|-------------|------------|
| **Interactivity** | ✅ Interactive dashboards | ❌ Static plots |
| **Real-time** | ✅ Live monitoring during training | ❌ Post-training only |
| **Publication** | ❌ Web-based only | ✅ High-quality static images |
| **Setup** | 📦 Requires installation | 🏠 Usually pre-installed |
| **Use Case** | 🔬 Experiment monitoring | 📑 Reports and papers |

### Best Practices

1. **Use both approaches**: TensorBoard for development, Matplotlib for reporting
2. **Organize logs**: Use timestamped directories for TensorBoard logs
3. **Memory management**: Close matplotlib figures when saving to prevent memory leaks
4. **Error handling**: Always include fallback options for missing dependencies
5. **Clear labeling**: Use descriptive titles and labels for better understanding

# 🔌 Ready-to-Use: Complete Training Visualizer

The following cell contains a complete, plug-and-play function that combines all the functionality demonstrated above. Simply copy this function into your own projects and use it with your real training data!

````python
def complete_training_visualizer(train_loss, val_loss, experiment_name="training_run", 
                                save_plot=True, show_plot=True, use_tensorboard=True):
    """
    🔌 PLUG-AND-PLAY COMPLETE TRAINING VISUALIZER
    
    A comprehensive function that handles all aspects of ML training visualization:
    - TensorBoard logging (if available and requested)
    - CSV fallback logging
    - Static plot generation and saving
    - Interactive plot display
    
    Args:
        train_loss (list): Training loss values per epoch
        val_loss (list): Validation loss values per epoch
        experiment_name (str): Name for your experiment (used in filenames and titles)
        save_plot (bool): Whether to save the plot as PNG file
        show_plot (bool): Whether to display the plot interactively
        use_tensorboard (bool): Whether to attempt TensorBoard logging
    
    Returns:
        dict: Summary of created files and paths
    
    Example Usage:
        # In your training loop:
        train_losses = []
        val_losses = []
        
        for epoch in range(num_epochs):
            # ... your training code ...
            train_losses.append(train_loss)
            val_losses.append(val_loss)
        
        # Visualize results:
        results = complete_training_visualizer(
            train_losses, val_losses, 
            experiment_name="my_model_v1"
        )
    """
    import matplotlib.pyplot as plt
    import os
    import time
    import csv
    
    # Try to import TensorBoard
    try:
        from torch.utils.tensorboard import SummaryWriter
        tensorboard_available = True
    except ImportError:
        tensorboard_available = False
    
    print(f"🚀 Starting visualization for: {experiment_name}")
    print(f"📊 Data: {len(train_loss)} epochs")
    
    # Initialize result tracking
    results = {
        "experiment_name": experiment_name,
        "epochs": len(train_loss),
        "files_created": [],
        "tensorboard_available": tensorboard_available
    }
    
    # === TENSORBOARD LOGGING ===
    writer = None
    log_dir = None
    
    if use_tensorboard and tensorboard_available:
        try:
            log_dir = os.path.join("logs", f"{experiment_name}_{int(time.time())}")
            writer = SummaryWriter(log_dir)
            
            print(f"📈 Logging to TensorBoard: {log_dir}")
            
            # Log all metrics
            for epoch, (t_loss, v_loss) in enumerate(zip(train_loss, val_loss), 1):
                writer.add_scalar('Loss/train', t_loss, epoch)
                writer.add_scalar('Loss/validation', v_loss, epoch)
                writer.add_scalars('Loss/comparison', {
                    'train': t_loss, 
                    'validation': v_loss
                }, epoch)
            
            writer.close()
            results["tensorboard_dir"] = log_dir
            results["files_created"].append(f"TensorBoard logs: {log_dir}")
            print(f"✅ TensorBoard logging complete!")
            
        except Exception as e:
            print(f"⚠️ TensorBoard logging failed: {e}")
    
    # === CSV LOGGING (ALWAYS) ===
    csv_filename = f"{experiment_name}_metrics.csv"
    try:
        with open(csv_filename, 'w', newline='') as csvfile:
            writer_csv = csv.writer(csvfile)
            writer_csv.writerow(['Epoch', 'Training_Loss', 'Validation_Loss'])
            
            for epoch, (t_loss, v_loss) in enumerate(zip(train_loss, val_loss), 1):
                writer_csv.writerow([epoch, t_loss, v_loss])
        
        results["csv_file"] = csv_filename
        results["files_created"].append(f"CSV data: {csv_filename}")
        print(f"📁 Metrics saved to: {csv_filename}")
        
    except Exception as e:
        print(f"⚠️ CSV logging failed: {e}")
    
    # === MATPLOTLIB PLOTTING ===
    epochs = range(1, len(train_loss) + 1)
    
    plt.figure(figsize=(12, 7))
    
    # Plot with enhanced styling
    plt.plot(epochs, train_loss, 'bo-', label='Training Loss', 
             linewidth=2.5, markersize=7, alpha=0.8)
    plt.plot(epochs, val_loss, 'ro-', label='Validation Loss', 
             linewidth=2.5, markersize=7, alpha=0.8)
    
    # Enhanced plot styling
    plt.title(f'{experiment_name}: Training vs Validation Loss', 
              fontsize=18, fontweight='bold', pad=20)\n    plt.xlabel('Epochs', fontsize=14, fontweight='semibold')\n    plt.ylabel('Loss', fontsize=14, fontweight='semibold')\n    plt.legend(fontsize=12, framealpha=0.9, shadow=True)\n    plt.grid(True, alpha=0.3, linestyle='--')\n    \n    # Add annotations for overfitting detection\n    if len(val_loss) > 3:\n        min_val_idx = val_loss.index(min(val_loss))\n        if min_val_idx < len(val_loss) - 2:  # Check if minimum is not at the end\n            plt.annotate('Potential Overfitting Zone', \n                        xy=(min_val_idx + 2, val_loss[min_val_idx + 1]), \n                        xytext=(min_val_idx + 3, max(val_loss) * 0.8),\n                        arrowprops=dict(arrowstyle='->', color='red', alpha=0.7),\n                        fontsize=10, color='red', alpha=0.8)\n    \n    plt.tight_layout()\n    \n    # Save plot if requested\n    if save_plot:\n        plot_filename = f\"{experiment_name}_loss_curve.png\"\n        try:\n            plt.savefig(plot_filename, dpi=300, bbox_inches='tight', \n                       facecolor='white', edgecolor='none')\n            results[\"plot_file\"] = plot_filename\n            results[\"files_created\"].append(f\"Plot image: {plot_filename}\")\n            print(f\"🎨 Plot saved to: {plot_filename}\")\n        except Exception as e:\n            print(f\"⚠️ Plot saving failed: {e}\")\n    \n    # Show plot if requested\n    if show_plot:\n        plt.show()\n    else:\n        plt.close()\n    \n    # === SUMMARY ===\n    print(\"\\n\" + \"=\"*50)\n    print(f\"🎉 Visualization Complete for: {experiment_name}\")\n    print(\"=\"*50)\n    print(f\"📊 Processed {len(train_loss)} epochs\")\n    print(f\"📈 Final Training Loss: {train_loss[-1]:.4f}\")\n    print(f\"📉 Final Validation Loss: {val_loss[-1]:.4f}\")\n    print(f\"🎯 Best Validation Loss: {min(val_loss):.4f} (Epoch {val_loss.index(min(val_loss)) + 1})\")\n    \n    print(\"\\n📁 Files Created:\")\n    for file_info in results[\"files_created\"]:\n        print(f\"   ✅ {file_info}\")\n    \n    if log_dir:\n        print(f\"\\n🌐 View TensorBoard: tensorboard --logdir={log_dir}\")\n    \n    return results

# 🧪 EXAMPLE USAGE\nprint(\"\\n\" + \"=\"*60)\nprint(\"🧪 DEMONSTRATION: Complete Training Visualizer\")\nprint(\"=\"*60)\n\n# Example with the same overfitting data\nexample_train_loss = [2.5, 1.8, 1.2, 0.9, 0.7, 0.5, 0.4, 0.3, 0.2]\nexample_val_loss = [2.7, 2.1, 1.6, 1.3, 1.1, 1.0, 1.1, 1.2, 1.4]\n\n# Call the complete visualizer\nresults = complete_training_visualizer(\n    example_train_loss, \n    example_val_loss,\n    experiment_name=\"demo_overfitting_model\",\n    save_plot=True,\n    show_plot=True,\n    use_tensorboard=True\n)\n\nprint(f\"\\n📋 Results Summary: {results}\")
````
